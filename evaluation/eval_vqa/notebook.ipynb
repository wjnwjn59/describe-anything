```python
import json
path = "/mnt/VLAI_data/InfographicVQA/Task 3/gt/infographicVQA_val_v1.0.json"
import os

with open(path, "r") as f:
    data = json.load(f)["data"]

new_data = []
for item in data:
    new_item = {
        "image": item["image_local_name"],
        "question": item["question"],
        "question_id": item["questionId"],
        "answer": item["answers"]
    }
    new_data.append(new_item)

with open("/mnt/VLAI_data/InfographicVQA/infographicvqa_val.jsonl", "w") as f:
    for item in new_data:
        f.write(json.dumps(item) + "\n")
```
